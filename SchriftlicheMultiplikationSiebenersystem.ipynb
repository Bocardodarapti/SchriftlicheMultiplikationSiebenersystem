Schriftliche Multiplikation im Siebenersystem

Dieses Werkzeug visualisiert die schriftliche Multiplikation im Siebenersystem.

```python
import time
from IPython.display import clear_output

def to_base_7(n):
    """Hilfsfunktion zur Umwandlung einer Dezimalzahl in Basis 7."""
    if n == 0: return "0"
    digits = []
    while n:
        digits.append(str(n % 7))
        n //= 7
    return "".join(digits[::-1])

def schriftliche_multiplikation_base7(zahl1_s, zahl2_s, v):
    # Validierung der Eingabe (nur Ziffern 0-6)
    for s in [zahl1_s, zahl2_s]:
        if not all(d in "0123456" for d in s):
            print("Fehler: Im Siebenersystem sind nur Ziffern von 0 bis 6 erlaubt!")
            return

    # Zeitsteuerung
    zeit_werte = [10, 7, 5, 4, 3, 2, 1.5, 1, 0.5, 0.1]
    v_clamped = max(1, min(10, v))
    delay = zeit_werte[v_clamped - 1]
    
    s1, s2 = zahl1_s, zahl2_s
    len1, len2 = len(s1), len(s2)
    
    # Berechnung des Endergebnisses zur Bestimmung der Breite
    val1_dec = int(s1, 7)
    val2_dec = int(s2, 7)
    final_res_s = to_base_7(val1_dec * val2_dec)
    
    block_width = max(len(final_res_s), len1 + len2)
    margin = 6 
    total_width = margin + block_width
    
    header_left_part_width = total_width - len(s2) - 3
    header_left = s1.rjust(header_left_part_width)
    header_mid = " * "
    header_right = s2.rjust(len(s2)) 
    header_line = header_left + header_mid + header_right
    
    teilprodukte = []
    
    # --- START: NUR HEADER ANZEIGEN ---
    clear_output(wait=True)
    print("Vorbereitung der Rechnung im Siebenersystem (Basis 7)...")
    print("-" * (total_width + 2))
    print(header_line)
    print("-" * (total_width + 2))
    time.sleep(delay)

    # --- PHASE 1: MULTIPLIKATION ---
    for i in range(len2 - 1, -1, -1):
        produkt_nummer = len2 - i
        ziffer_unten_dec = int(s2[i])
        
        # Multiplikation im Siebenersystem
        prod_dec = int(s1, 7) * ziffer_unten_dec
        p_str = to_base_7(prod_dec)
        teilprodukte.append(p_str)
        
        clear_output(wait=True)
        print(f"Phase 1: Multiplikation | {produkt_nummer}. Teilprodukt wird berechnet...")
        print("-" * (total_width + 2))
        print(header_line)
        print("-" * (total_width + 2))
        
        for idx, p in enumerate(teilprodukte):
            print(p.rjust(total_width - idx))
            
        print("-" * (total_width + 2))
        time.sleep(delay)

    # --- PHASE 2: ADDITION (Basis 7) ---
    matrix = [[0] * block_width for _ in range(len(teilprodukte))]
    for idx, p in enumerate(teilprodukte):
        p_digits = [int(d) for d in p]
        for j in range(len(p_digits)):
            pos = block_width - 1 - idx - j
            if pos >= 0:
                matrix[idx][pos] = p_digits[-(j+1)]

    ergebnis_liste = [" "] * block_width
    uebertraege_liste = [" "] * block_width
    
    uebertrag = 0
    for i in range(block_width - 1, -1, -1):
        spalten_summe = uebertrag
        for row in matrix:
            spalten_summe += row[i]
        
        # Basis 7 Logik
        digit = spalten_summe % 7
        ergebnis_liste[i] = str(digit)
        
        neuer_uebertrag = spalten_summe // 7
        if i > 0:
            uebertraege_liste[i-1] = str(neuer_uebertrag) if neuer_uebertrag > 0 else " "
        
        clear_output(wait=True)
        print(f"Phase 2: Spaltenweise Addition (Basis 7) | Stelle {block_width - i}")
        print("-" * (total_width + 2))
        print(header_line)
        print("-" * (total_width + 2))
        for idx, p in enumerate(teilprodukte):
            print(p.rjust(total_width - idx))
        
        uebertrag_str = "".join(uebertraege_liste).rjust(total_width)
        ergebnis_str = "".join(ergebnis_liste).rjust(total_width)
        
        print(uebertrag_str)
        print("-" * (total_width + 2))
        print(ergebnis_str)
        print(" " * (total_width - 1 - (block_width - 1 - i)) + "^")
        
        uebertrag = neuer_uebertrag
        time.sleep(delay)

    # --- FINALER ABSCHLUSS ---
    clear_output(wait=True)
    print("Dokumentation der Gesamtrechnung im Siebenersystem (Basis 7):")
    print("-" * (total_width + 2))
    print(header_line)
    print("-" * (total_width + 2))
    for idx, p in enumerate(teilprodukte):
        print(p.rjust(total_width - idx))
    print("-" * (total_width + 2))
    print(final_res_s.rjust(total_width))
    print("-" * (total_width + 2))
```

```python
# Eingabe der Zahlen als Strings, da sie im Siebersystem stehen
zahl1 = "163" # Beispiel Basis 7
zahl2 = "405"  # Beispiel Basis 7
v = 5         # Geschwindigkeit

schriftliche_multiplikation_base7(zahl1, zahl2, v)
```
